```python
import os
import shutil

DATASET_PATH = "/kaggle/input/datasets/asthem014/mmrdr-dr-detection"
WORK_DIR = "/kaggle/working/MMRDR"

os.makedirs(WORK_DIR, exist_ok=True)

combined_zip = os.path.join(WORK_DIR, "MMRDR.zip")

with open(combined_zip, "wb") as outfile:
    for i in range(1, 10):
        part = os.path.join(DATASET_PATH, f"MMRDR.zip.{i:03d}")
        print(f"Adding part {i}/9...")
        
        with open(part, "rb") as infile:
            shutil.copyfileobj(infile, outfile)

print("\nMMRDR.zip created successfully!")
print(f"Size: {os.path.getsize(combined_zip)/(1024**3):.2f} GB")
```

```python
import subprocess
import os

DATASET_PATH = "/kaggle/input/datasets/asthem014/mmrdr-dr-detection"
OUTPUT_PATH = "/kaggle/working/MMRDR-CFP"

os.makedirs(OUTPUT_PATH, exist_ok=True)

cmd = [
    "7z", "x",
    f"{DATASET_PATH}/MMRDR.zip.001",
    "MMRDR-CFP/*",
    "-o" + OUTPUT_PATH,
    "-y"
]

print("Starting CFP extraction...")
print("This may take several minutes.\n")

result = subprocess.run(cmd)

print("\nExtraction finished.")
print("Return code:", result.returncode)
```
